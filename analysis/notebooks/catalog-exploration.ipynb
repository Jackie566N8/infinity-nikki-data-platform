# Catalog evidence and exploratory analysis
Run the maintained build and `analysis/research.py` first. This notebook reads saved outputs; it does not tune frozen search, collect personal data or replace human Gold review.

```python
from pathlib import Path
import json
import pandas as pd
import duckdb
ROOT = next(base for base in (Path.cwd(), *Path.cwd().parents) if (base / "data/dataset").exists())
quality = json.loads((ROOT / "data/dataset/quality-report.json").read_text(encoding="utf-8"))
pd.DataFrame(quality["metrics"])[["key", "numerator", "denominator", "percent", "definition"]]
```

```python
con = duckdb.connect(str(ROOT / "data/dataset/research/research.duckdb"), read_only=True)
con.sql("SELECT style, rarity, count(*) AS n FROM objects JOIN outfits ON object_id = outfit_id WHERE is_base = 1 GROUP BY style, rarity ORDER BY style, rarity").df()
```

## Interpretation
Counts cover the indexed catalog. They do not measure player preferences, and collection dates are not release dates. Inspect sparse cells and provenance before making any trend claim.
